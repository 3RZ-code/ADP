<b>Ćwiczenie 9.</b> Zamień liczbę minut na przedziały czasowe o wartościach 5, 10, 15, ..., 50, 55, 60. Zapisz rezultaty w kolumnach <i>pt_r_t_cl_Itis</i>, <i>car_r_t_cl_Itis</i>, <i>pt_r_t_cl_Myyrmanni</i> oraz <i>pt_r_t_cl_Myyrmanni</i>.

```python
import mapclassify

classifier = mapclassify.UserDefined.make(bins = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60])

### your code ###

grid["pt_r_t_cl_Itis"] = classifier(grid["pt_r_t_Itis"]) * 5
grid["car_r_t_cl_Itis"] = classifier(grid["car_r_t_Itis"]) * 5
grid["pt_r_t_cl_Myyrmanni"] = classifier(grid["pt_r_t_Myyrmanni"]) * 5
grid["car_r_t_cl_Myyrmanni"] = classifier(grid["car_r_t_Myyrmanni"]) * 5

grid
```
